```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Sneha","Arjun","Kiran","Pooja","Vijay"],
    "Department":["IT","HR","IT","Sales","HR","Sales","IT","HR"],
    "Age":[25,28,30,24,29,31,27,26],
    "Salary":[50000,45000,60000,40000,52000,48000,65000,47000],
    "City":["Delhi","Mumbai","Delhi","Chennai","Hyderabad","Mumbai","Delhi","Chennai"],
    "Experience":[2,4,6,1,5,7,3,2]
})
```

```python
df
```

```python
# Exploring dataset

print("First 5 rows:\n")
print(df.head())
print("-------------------------------------------------------------------")
print("Last 3 rows:\n")
print(df.tail(3))
print("-------------------------------------------------------------------")
print("Shape of the DataFrame:\n")
print(df.shape)
print("-------------------------------------------------------------------")
print("Summary statistics:\n")
print(df.describe())
print("-------------------------------------------------------------------")
print("DataFrame Information:\n")
print(df.info())
```

```python
# Data Selection

print("Name and Salary:\n")
print(df[["Name","Salary"]])
print("-------------------------------------------------------------------")
print("First Employee:\n")
print(df.head(1))
print("-------------------------------------------------------------------")
print("employees from 2 to 5:\n")
print(df[2:6])
print("-------------------------------------------------------------------")
print("Salary of Rahul:\n")
print(df[df["Name"]=="Rahul"]["Salary"])
print("-------------------------------------------------------------------")
print("City of employee 105:\n")
print(df[df["Emp_ID"]==105]["City"])
```

```python
#Filtering Data

print("employees salary greater than 50k:\n")
print(df[df["Salary"]>50000])
print("-------------------------------------------------------------------")
print("employees from Delhi:\n")
print(df[df["City"] == "Delhi"])
print("-------------------------------------------------------------------")
print("employees from HR dept.:\n")
print(df[df["Department"] == "HR"])
print("-------------------------------------------------------------------")
print("employees age who are greater than 25:\n")
print(df[df["Age"]>27])
print("-------------------------------------------------------------------")
print("employees who's salary between 45k to 55k:\n")
print(df[df["Salary"].between(45000,55000)])
print("-------------------------------------------------------------------")
print("employees from delhi or mumbai:\n")
print(df[df["City"].isin(["Delhi","Mumbai"])])
print("-------------------------------------------------------------------")
print("employee from it dept experience is greatter than 2:\n")
print(df[(df["Department"]=="HR") & (df["Experience"]>2)])
print("-------------------------------------------------------------------")
print("employees whose name is priya or rahul:\n")
print(df[df["Name"].isin(["Priya","Rahul"])])
```

```python
# Data Cleaning

print("Checking for Null values:\n")
print(df.isna())
print("-------------------------------------------------------------------")
print("Number of Missing Values:\n")
print(df.isna().sum())
print("-------------------------------------------------------------------")
print("Checking for duplicate values:\n")
print(df.duplicated())
print("-------------------------------------------------------------------")
print("if any duplicates are these removing them:\n")
print(df.drop_duplicates())
```

```python
# Data Transformation
# "Salary after bonus of 10%
df["Bonus"] = df["Salary"].apply(lambda x: x*0.10)
# Total Salary
df["Total Salary"] = df["Salary"]+df["Bonus"]
# converting cities to Upper case
df["City"] = df["City"].str.upper()
# Replacing delhi with new delhi
df.replace("DELHI","NEW DELHI",inplace=True)
# Renaming column salary to montly salary
df.rename(columns={"Salary":"Montly Salary"},inplace=True)
df
```

```python
# Sorting
df.sort_values(["Department","Montly Salary"],ascending=[True,True])
```

```python
# Groupby

print("\n#avg salary in each dept.\n")
print(df.groupby("Department")["Montly Salary"].mean())
print("\n#max salary in each dept.\n")
print(df.groupby("Department")["Montly Salary"].max())
print("\n#min salary in each dept.\n")
print(df.groupby("Department")["Montly Salary"].min())
print("\n#total salary in each dept.\n")
print(df.groupby("Department")["Montly Salary"].sum())
print("\n#no of employees in each dept.\n")
print(df.groupby("Department")["Emp_ID"].count())
print("\n#Avg exp in eacj dept.\n")
print(df.groupby("Department")["Experience"].mean())
```

```python
melted_data = df.melt(
    id_vars=["Emp_ID", "Name", "Department"],
    value_vars=["Age", "Montly Salary", "Experience"],
    var_name="Attribute",
    value_name="Value"
)

melted_data
```

```python
#pivot
pivot_data = df.pivot_table(
    values="Montly Salary",
    index="Department",
    columns="City",
    aggfunc="mean"
)

pivot_data
```

```python
#merging
department_info = pd.DataFrame({
    "Department": ["IT", "HR", "Sales"],
    "Manager": ["Anil", "Kavya", "Rohit"],
    "Location": ["Bangalore", "Hyderabad", "Chennai"]
})

df = pd.merge(df, department_info, on="Department", how="left")

df
```

```python
# saving data to csv file

df.to_csv("Employee.csv")
```
